# Torino and Palermo scales

Lore F9. Torino reads the (P, E) chart; Palermo compares the risk with the background rate over the time left.

Uses the Python mirror `pymiss` (same formulas and names as `src/physics`).

```python
import os, sys, math
# Find the repo root (the folder that holds pymiss/) whether the notebook runs from the root or from pymiss/notebooks.
root = os.getcwd()
while not os.path.isdir(os.path.join(root, 'pymiss')) and os.path.dirname(root) != root:
    root = os.path.dirname(root)
sys.path.insert(0, root)
from pymiss.physics.risk import torino, torino_band, palermo, background_frequency
cases = [('2024 YR4 (Feb 2025)', 0.031, 8, 7), ('Apophis (Dec 2004)', 0.027, 500, 25), ('2023 DW', 1.7e-3, 9, 23), ('a certain global hit', 1.0, 1e6, 10)]
for name, P, E, T in cases:
    lvl = torino(P, E)
    print(f'{name:22s} Torino {lvl} {torino_band(lvl):6s}  f_B {background_frequency(E):.2e}/yr  PS {palermo(P, E, T):+.2f}')
```
