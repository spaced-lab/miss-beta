# Gravity assist: turn without fuel

Uses the Python mirror `pymiss` (same formulas and names as `src/physics`). Run from the repo root.


```python
import sys, math
sys.path.insert(0, '..' if 'notebooks' in __import__('os').getcwd() else '.')
from pymiss.physics.constants import AU, MU_SUN
from pymiss.physics.flyby import turn_angle, flyby
from pymiss.physics.kepler import vis_viva
mu, R, a = 1.26686534e17, 6.9911e7, 5.203 * AU   # GIANT (src/config/bodies.json)
v_inf, r_p = 10e3, 10 * R
delta = turn_angle(r_p, v_inf, mu)
print('turn angle (deg):', math.degrees(delta))
vp = vis_viva(MU_SUN, a, a)
phi = math.radians(120)
vpx, vpy = vp * math.cos(phi), vp * math.sin(phi)
for side in (1, -1):
    ox, oy, vx, vy, dv, rp = flyby(vpx + v_inf, vpy, vpx, vpy, delta, side, mu)
    print('side', side, 'dv_helio (km/s):', dv / 1e3, ' sun-frame speed in -> out (km/s):', math.hypot(vpx + v_inf, vpy) / 1e3, math.hypot(vx, vy) / 1e3)
```
