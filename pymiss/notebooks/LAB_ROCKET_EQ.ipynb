# Rocket equation and payload vs C3

Lore D8. dv = Isp g0 ln(m0/mf); staging multiplies it; payload falls steeply with C3.

Uses the Python mirror `pymiss` (same formulas and names as `src/physics`).

```python
import os, sys, math
# Find the repo root (the folder that holds pymiss/) whether the notebook runs from the root or from pymiss/notebooks.
root = os.getcwd()
while not os.path.isdir(os.path.join(root, 'pymiss')) and os.path.dirname(root) != root:
    root = os.path.dirname(root)
sys.path.insert(0, root)
from pymiss.physics.rocket import rocket_dv, staged_dv, mass_ratio_for
from pymiss.physics.estimates import LAUNCHERS, payload_for_c3
print('one stage, Isp 330 s, f 0.88:', rocket_dv(330, 1 / (1 - 0.88)) / 1000, 'km/s')
for n in (1, 2, 3, 4):
    dv = staged_dv(330, 0.88, n)
    print(f'{n} stages: {dv / 1000:.2f} km/s, overall mass ratio {mass_ratio_for(dv, 330):.1f}')
for c3 in (0, 20, 40, 60, 90):
    print(f'C3 {c3:3d}:', ', '.join(f"{t['name']} {payload_for_c3(t, c3) / 1000:.2f} t" for t in LAUNCHERS))
```
