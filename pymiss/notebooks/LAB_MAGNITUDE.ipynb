# Brightness: H and V

Uses the Python mirror `pymiss` (same formulas and names as `src/physics`). Run from the repo root.


```python
import sys, math
sys.path.insert(0, '../..' if 'notebooks' in __import__('os').getcwd() else '.')
from pymiss.physics.constants import AU, MU_SUN, DAY, YEAR
from pymiss.physics.photometry import absolute_magnitude, apparent_magnitude
H = absolute_magnitude(140, 0.14)
print('H:', H)
for delta in (0.01, 0.05, 0.1, 0.5):
    print(delta, 'AU -> V', apparent_magnitude(H, 1.1, delta, 20))
```
