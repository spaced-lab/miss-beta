# The ellipse: a, e, w

Uses the Python mirror `pymiss` (same formulas and names as `src/physics`). Run from the repo root.


```python
import sys, math
sys.path.insert(0, '../..' if 'notebooks' in __import__('os').getcwd() else '.')
from pymiss.physics.constants import AU, MU_SUN, DAY, YEAR
from pymiss.physics.kepler import Elements, propagate, period, periapsis, apoapsis
el = Elements(1.3 * AU, 0.4, math.radians(30), math.radians(60), 0.0)
print('q, Q (AU):', periapsis(el) / AU, apoapsis(el) / AU)
print('T (d):', period(MU_SUN, el.a) / DAY)
x, y, vx, vy = propagate(el, MU_SUN, 0.0)
print('r (AU), v (km/s):', math.hypot(x, y) / AU, math.hypot(vx, vy) / 1000)
```
