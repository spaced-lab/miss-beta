# Kepler solver: Newton on E - e sin E = M

Uses the Python mirror `pymiss` (same formulas and names as `src/physics`). Run from the repo root.


```python
import sys, math
sys.path.insert(0, '../..' if 'notebooks' in __import__('os').getcwd() else '.')
from pymiss.physics.constants import AU, MU_SUN, DAY, YEAR
from pymiss.physics.kepler import solve_kepler
e, M = 0.8, 1.0
E = math.pi
for i in range(6):
    f = E - e * math.sin(E) - M
    E -= f / (1 - e * math.cos(E))
    print(i + 1, E, f)
print('library:', solve_kepler(M, e))
```
