# Solar sail: the slow free push

Uses the Python mirror `pymiss` (same formulas and names as `src/physics`). Run from the repo root.


```python
import sys, math
sys.path.insert(0, '..' if 'notebooks' in __import__('os').getcwd() else '.')
from pymiss.physics.constants import AU, MU_SUN, YEAR
from pymiss.physics.sail import characteristic_accel, lightness, integrate_sail
for sigma_g in (100, 25, 10, 3):
    a_c = characteristic_accel(sigma_g / 1000, 0.9)
    s0 = (AU, 0.0, 0.0, math.sqrt(MU_SUN / AU))
    state, dv = integrate_sail(s0, 0.0, YEAR, a_c, MU_SUN, 'SPIRAL_OUT')
    print('sigma %3d g/m2: a_c %.3f mm/s2, beta_s %.3f, one year: dv %.2f km/s, r %.2f AU' % (sigma_g, a_c * 1e3, lightness(a_c), dv / 1e3, math.hypot(state[0], state[1]) / AU))
```
