# Euler vs RK4: why the method matters

Uses the Python mirror `pymiss` (same formulas and names as `src/physics`). Run from the repo root.


```python
import sys, math
sys.path.insert(0, '..' if 'notebooks' in __import__('os').getcwd() else '.')
from pymiss.physics.constants import AU, MU_SUN
from pymiss.physics.integrate import integrate, euler_step, rk4_step, gravity_deriv
from pymiss.physics.kepler import period, specific_energy
e = 0.5
rp = AU * (1 - e)
s0 = (rp, 0.0, 0.0, math.sqrt(MU_SUN * (1 + e) / rp))
E0 = specific_energy(MU_SUN, rp, s0[3])
T = period(MU_SUN, AU)
for h_hours in (64, 32, 16, 8, 4):
    errs = []
    for step in (euler_step, rk4_step):
        s = integrate(gravity_deriv(MU_SUN), s0, 0.0, T, h_hours * 3600, step)
        E = specific_energy(MU_SUN, math.hypot(s[0], s[1]), math.hypot(s[2], s[3]))
        errs.append(abs((E - E0) / E0))
    print('h = %3d h: energy error Euler %.2e, RK4 %.2e' % (h_hours, errs[0], errs[1]))
```
