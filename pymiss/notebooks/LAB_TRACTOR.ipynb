# Slow push: tractor, ion beam, laser, mass driver

Lore D6. Millimetres per second per year: miss = 3 a tau (T_rem + tau/2).

Uses the Python mirror `pymiss` (same formulas and names as `src/physics`).

```python
import os, sys, math
# Find the repo root (the folder that holds pymiss/) whether the notebook runs from the root or from pymiss/notebooks.
root = os.getcwd()
while not os.path.isdir(os.path.join(root, 'pymiss')) and os.path.dirname(root) != root:
    root = os.path.dirname(root)
sys.path.insert(0, root)
from pymiss.physics.slow_push import gravity_tractor_force, push_force
from pymiss.physics.estimates import miss_from_push
from pymiss.physics.close_approach import sphere_mass
from pymiss.physics.constants import YEAR, R_HOME
D = 100.0
M, R = sphere_mass(D, 2700), D / 2
tau, t_rem = 5 * YEAR, 5 * YEAR
forces = {'GRAVITY_TRACTOR (20 t, 1.5 R)': gravity_tractor_force(20000, M, R, 1.5)}
for meth in ('ION_BEAM', 'LASER', 'MASS_DRIVER'):
    forces[meth + ' tier 1'] = push_force(meth, 1, M, R)
for name, F in forces.items():
    a = F / M
    print(f'{name:30s} F {F:.3g} N  dv {1000 * a * tau:.3g} mm/s  miss {miss_from_push(a, tau, t_rem) / R_HOME:.3g} Earth radii')
```
