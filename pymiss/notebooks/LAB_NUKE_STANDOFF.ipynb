# Nuclear standoff: X-ray ablation

Lore D5. The intercepted fraction f_omega falls fast with burst height; the push is dp = C_m E_abs.

Uses the Python mirror `pymiss` (same formulas and names as `src/physics`).

```python
import os, sys, math
# Find the repo root (the folder that holds pymiss/) whether the notebook runs from the root or from pymiss/notebooks.
root = os.getcwd()
while not os.path.isdir(os.path.join(root, 'pymiss')) and os.path.dirname(root) != root:
    root = os.path.dirname(root)
sys.path.insert(0, root)
from pymiss.physics.nuclear import nuclear_burst, intercepted_fraction
from pymiss.physics.fragment import q_star
from pymiss.physics.close_approach import sphere_mass
R, rho, Y = 500.0, 2700, 1.0
M = sphere_mass(2 * R, rho)
for hr in (0.1, 0.25, 0.5, 1, 2, 5):
    e_abs, dp, dv, Q, f = nuclear_burst(Y, 'STANDOFF', M, R, hr * R)
    print(f'H/R {hr:4}: f_omega {f:.3f}  dv {100 * dv:.3g} cm/s  Q {Q:.3g} vs Q* {q_star(R, rho):.3g} J/kg')
```
