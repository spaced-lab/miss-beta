# Disruption threshold Q* versus size

Lore T5, D4. Strength holds small bodies together, gravity holds big ones: Q* has a minimum near a few hundred metres.

Uses the Python mirror `pymiss` (same formulas and names as `src/physics`).

```python
import os, sys, math
# Find the repo root (the folder that holds pymiss/) whether the notebook runs from the root or from pymiss/notebooks.
root = os.getcwd()
while not os.path.isdir(os.path.join(root, 'pymiss')) and os.path.dirname(root) != root:
    root = os.path.dirname(root)
sys.path.insert(0, root)
from pymiss.physics.fragment import q_star, specific_energy
from pymiss.physics.close_approach import sphere_mass
m, U = 1000, 10000  # impactor kg, m/s
for R in (1, 10, 100, 150, 300, 1000, 5000):
    Qs = q_star(R, 2700)
    Q = specific_energy(m, sphere_mass(2 * R, 2700), U)
    print(f'R {R:5d} m  Q* {Qs:9.3g} J/kg  Q {Q:9.3g}  Q/Q* {Q / Qs:8.3g}  {"DISRUPTED" if Q >= Qs else "intact"}')
```
