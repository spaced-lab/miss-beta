# Impact energy: E = 1/2 M v^2

Lore T2. Mass grows as D^3, so doubling the size gives eight times the energy.

Uses the Python mirror `pymiss` (same formulas and names as `src/physics`).

```python
import os, sys, math
# Find the repo root (the folder that holds pymiss/) whether the notebook runs from the root or from pymiss/notebooks.
root = os.getcwd()
while not os.path.isdir(os.path.join(root, 'pymiss')) and os.path.dirname(root) != root:
    root = os.path.dirname(root)
sys.path.insert(0, root)
from pymiss.physics.close_approach import escape_speed, impact_energy, impact_category
from pymiss.physics.constants import MU_HOME, R_HOME
v_esc = escape_speed(MU_HOME, R_HOME)
v = math.sqrt(17000**2 + v_esc**2)
for D in (20, 50, 140, 1000, 10000):
    m, J, Mt = impact_energy(D, 2700, v)
    print(f'D {D:6d} m  M {m:.2e} kg  E {Mt:.3g} Mt  {impact_category(D)}  = {Mt / 0.015:.3g} Hiroshimas')
```
